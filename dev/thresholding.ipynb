```python
import glob
import cv2
import matplotlib.pyplot as plt
import numpy as np
```

```python
# Input
folder_in_main = '/media/joris/rootfs/home/joris/data/kiwibes2026/20260706kiwibes/bes21-30-joris'
```

```python
# Loop over berries
folder_in_berry = f'{folder_in_main}/bes21'

# Loop over images
file_in = glob.glob(f'{folder_in_berry}/*')[0]
```

```python
file_in = np.random.choice(glob.glob(f'{folder_in_main}/*/*'))
print(file_in)

# Load
img = cv2.imread(file_in)

# Blur
blur = cv2.GaussianBlur(img, (15, 15), 0)  

# Convert to grayscale
gray = cv2.cvtColor(blur, cv2.COLOR_BGR2GRAY)

# Applying threshold
threshold = cv2.threshold(gray, 0, 255, cv2.THRESH_BINARY_INV | cv2.THRESH_OTSU)[1]

# Connected components
analysis = cv2.connectedComponentsWithStats(threshold,
                                            4,
                                            cv2.CV_32S)
(totalLabels, label_ids, values, centroid) = analysis


# Threshold berry
th0 = 100
plt.figure(figsize=(20,10))
plt.subplot(131)
plt.imshow(img)
plt.axis('off')
plt.subplot(132)
plt.imshow(threshold)
plt.axis('off')
plt.subplot(133)
plt.axis('off')
plt.imshow(label_ids)
plt.axis('off')
plt.show()
```
